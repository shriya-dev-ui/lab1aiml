```python
# Road network using adjacency list

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A'],
    'D': ['B'],
    'E': ['F'],
    'F': ['E'],
    'G': []
}


# Function to check route using DFS
def route_exists(graph, start, end, visited=None):

    if visited is None:
        visited = set()

    if start == end:
        return True

    visited.add(start)

    for city in graph[start]:
        if city not in visited:
            if route_exists(graph, city, end, visited):
                return True

    return False


# Function to find connected components
def connected_components(graph):
    visited = set()
    components = []

    for city in graph:
        if city not in visited:

            component = []

            def dfs(node):
                visited.add(node)
                component.append(node)

                for neighbour in graph[node]:
                    if neighbour not in visited:
                        dfs(neighbour)

            dfs(city)
            components.append(component)

    return components


# Check route
start = 'A'
end = 'D'

if route_exists(graph, start, end):
    print("Route exists from", start, "to", end)
else:
    print("No route exists from", start, "to", end)


# Find connected components
components = connected_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, 1):
    print("Component", i, ":", component)
```
