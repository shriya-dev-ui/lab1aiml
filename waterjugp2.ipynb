```python
from collections import deque

def water_jug():
    queue = deque([((0, 0), [])])
    visited = set()

    while queue:
        (a, b), path = queue.popleft()

        if (a, b) in visited:
            continue

        visited.add((a, b))

        # Goal
        if a == 2 or b == 2:
            return path + [(a, b)]

        # Fill Jug A
        queue.append(((4, b), path + [(a, b)]))

        # Fill Jug B
        queue.append(((a, 3), path + [(a, b)]))

        # Empty Jug A
        queue.append(((0, b), path + [(a, b)]))

        # Empty Jug B
        queue.append(((a, 0), path + [(a, b)]))

        # Pour A -> B
        amount = min(a, 3 - b)
        queue.append(((a - amount, b + amount),
                      path + [(a, b)]))

        # Pour B -> A
        amount = min(b, 4 - a)
        queue.append(((a + amount, b - amount),
                      path + [(a, b)]))


# Call function
solution = water_jug()

print("Minimum Steps:")

for i in range(len(solution) - 1):
    print(i + 1, ".", solution[i], "->", solution[i + 1])

print("Goal:", solution[-1])
```
